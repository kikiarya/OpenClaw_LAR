```python
import json
from glob import glob
```

```python
with open('../dataset/kodcode_traj_qwen3/original_qwen3_sample.jsonl', 'w') as f:
    for file in glob('../dataset/kodcode_traj_qwen3/filtered_kodcode_qwen/sample*.json'):
        with open(file, 'r') as fin:
            data = json.load(fin)
            f.write(json.dumps({
                'id': f"triviaqa_{file.replace('.json', '').split('/')[-1].split('_')[1]}",
                'conversations': data['messages']
            }) + '\n')
```
