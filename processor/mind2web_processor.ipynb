```python
import json
from glob import glob
```

```python
with open('../dataset/mind2web_traj_llama_nothink/original_mind2web_sample.jsonl', 'w') as f:
    for file in glob('../dataset/mind2web_traj_llama_nothink/filtered_llama_traj_nothink/mind2web*.json'):
        print(file)
        with open(file, 'r') as fin:
            data = json.load(fin)
            f.write(json.dumps({
                'id': f"mind2web_{file.replace('.json', '').split('/')[-1].split('_')[1]}",
                'conversations': data['messages']
            }) + '\n')
```
